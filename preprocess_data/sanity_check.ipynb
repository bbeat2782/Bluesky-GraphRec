```python
import pandas as pd
import os
import logging

# # Setup logging configuration
# logging.basicConfig(level=logging.INFO)
# logger = logging.getLogger(__name__)

# post_embeddings_path = os.path.join(os.path.expanduser("~"), 'post_dynamic_embeddings.parquet')
# post_embeddings = pd.read_parquet(post_embeddings_path)
# logger.info(f"Successfully loaded {len(post_embeddings)} post embeddings")

import pickle
import os

# Define the path to your processed data
PROCESSED_DATA_PATH = "../processed_data/bluesky"  # Adjust path as needed

# Load user dynamic features
with open(os.path.join(PROCESSED_DATA_PATH, "user_dynamic_features_0.pkl"), "rb") as f:
    user_dynamic_features_0 = pickle.load(f)

# Load producer dynamic features
with open(os.path.join(PROCESSED_DATA_PATH, "producer_dynamic_features_0.pkl"), "rb") as f:
    producer_dynamic_features_0 = pickle.load(f)

# Load user dynamic features
with open(os.path.join(PROCESSED_DATA_PATH, "user_dynamic_features.pkl"), "rb") as f:
    user_dynamic_features = pickle.load(f)

# Load producer dynamic features
with open(os.path.join(PROCESSED_DATA_PATH, "producer_dynamic_features.pkl"), "rb") as f:
    producer_dynamic_features = pickle.load(f)
```

```python
# Check if user_dynamic_features and user_dynamic_features_0 are equivalent
print("Checking user_dynamic_features equivalence:")
user_dates = set(user_dynamic_features.keys())
user_dates_0 = set(user_dynamic_features_0.keys())
print(f"Same dates: {user_dates == user_dates_0}")

# Check a sample date if available
if len(user_dates) > 0:
    sample_date = list(user_dates)[0]
    if sample_date in user_dates_0:
        sample_users = set(user_dynamic_features[sample_date].keys())
        sample_users_0 = set(user_dynamic_features_0[sample_date].keys())
        print(f"Same users for date {sample_date}: {sample_users == sample_users_0}")
        
        # Check a sample user embedding if available
        if len(sample_users) > 0:
            sample_user = list(sample_users)[0]
            if sample_user in sample_users_0:
                emb_equal = (user_dynamic_features[sample_date][sample_user] == 
                             user_dynamic_features_0[sample_date][sample_user]).all()
                print(f"Same embedding for user {sample_user}: {emb_equal}")
                print(f"User embedding from user_dynamic_features:")
                print(user_dynamic_features[sample_date][sample_user])
                print(f"\nUser embedding from user_dynamic_features_0:")
                print(user_dynamic_features_0[sample_date][sample_user])

# Check if producer_dynamic_features and producer_dynamic_features_0 are equivalent
print("\nChecking producer_dynamic_features equivalence:")
producer_dates = set(producer_dynamic_features.keys())
producer_dates_0 = set(producer_dynamic_features_0.keys())
print(f"Same dates: {producer_dates == producer_dates_0}")

# Check a sample date if available
if len(producer_dates) > 0:
    sample_date = list(producer_dates)[0]
    if sample_date in producer_dates_0:
        sample_producers = set(producer_dynamic_features[sample_date].keys())
        sample_producers_0 = set(producer_dynamic_features_0[sample_date].keys())
        print(f"Same producers for date {sample_date}: {sample_producers == sample_producers_0}")
        
        # Check a sample producer embedding if available
        if len(sample_producers) > 0:
            sample_producer = list(sample_producers)[0]
            if sample_producer in sample_producers_0:
                emb_equal = (producer_dynamic_features[sample_date][sample_producer] == 
                             producer_dynamic_features_0[sample_date][sample_producer]).all()
                print(f"Same embedding for producer {sample_producer}: {emb_equal}")
                print(f"Producer embedding from producer_dynamic_features:")
                print(producer_dynamic_features[sample_date][sample_producer])
                print(f"\nProducer embedding from producer_dynamic_features_0:")
                print(producer_dynamic_features_0[sample_date][sample_producer])
```
